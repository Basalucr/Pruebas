## a) Sistema del túnel

```python
# --------------------- Librerias para todo el Notebook --------------------- #

import numpy as np
import math
import matplotlib.pyplot as plt
import copy
from scipy.constants import c # Librería con la constante de la velocidad de la luz #

# --------------------- Datos --------------------- #

L_0 = 180.0 # Longitud propia de la nave (m) #
D_0 = 120.0 # Longitud propia del tunel (m), es igual a la posición de salida en el sistema túnel # 
v_O= 0.800*c # Velocidad de la nave respecto al tunel (m/s) #
x_0_O= 0.0 # posición entrada (m) #
β= v_O / c  # Relación de velocidades # 
γ= 1 / math.sqrt(1-β**2) # Factor de Lorentz #

L_O =  L_0 / γ # La logitud de la nave se tiene que ajustar al sistema de referencia del túnel # 
x_i_punta_O = (D_0-L_O)/2 + L_O # Posición de la punta de la nave en t=0. Recordar que la nave está centrada en el túnel para t = 0 #
x_i_cola_O = (D_0-L_O)/2 # Posición de la cola de la nave en t=0. Recordar que la nave está centrada en el túnel para t = 0 #

x_i = -750 # Posición inicial x (m) (sistema túnel) para las curvas. Arbitraria por estética de las gráficas en a) y c) #
x_f = 750 # Posición final x (m) (sistema túnel) para las curvas. Arbitraria por estética de las gráficas en a) y c) #

ct_i = c*x_i/v_O  # Posición inicial ct (m) para las curvas #
ct_f = c*x_f/v_O  # Posición final ct (m) para las curvas #

# --------------------- Curvas --------------------- #

ct_O= np.linspace(ct_i, ct_f, 100) # Arreglo de ct para las curvas # 
x_entrada_O = np.full_like(ct_O,x_0_O) # Arreglo x para la línea de entrada # 
x_salida_O = np.full_like(ct_O,D_0) # Arreglo x para línea de salida # 

x_O = np.linspace(x_i, x_f, 100) # Arreglo de x para las curvas # 
ct_punta_O= c*(x_O - x_i_punta_O)/v_O  # Arreglo ct para la línea de punta #
ct_cola_O=  c*(x_O - x_i_cola_O)/v_O  # Arreglo ct para la línea de cola #

ct_luz_positiva_O = copy.copy(x_O)  # ct = x, arreglo ct para la recta de luz positiva. No lo solicitan pero es más sencillo obtenerlo aquí y luego hacer la transformación en b) #
ct_luz_negativa_O = -x_O # ct = -x, arreglo ct para la recta de luz negativa. No lo solicitan pero es más sencillo obtenerlo aquí y luego hacer la transformación en b) #

# --------------------- Puntos --------------------- #

A_x_O = x_0_O # Punto A, emisión en la entrada del túnel #
A_ct_O = c*(A_x_O - x_0_O)/v_O 

B_x_O = D_0 # Punto B, emisión en la salida del túnel #
B_ct_O = c*(B_x_O - D_0 )/v_O 

P_x_O = x_0_O # Punto P, paso de la punta por la entrada #
P_ct_O = c*(P_x_O - x_i_punta_O)/v_O # Recordar que se debe hacer un ajuste porque la nave está centrada en el túnel para t = 0 #

Q_x_O = D_0 # Punto Q, paso de la punta por la salida #
Q_ct_O = c*(Q_x_O - x_i_punta_O)/v_O # Recordar que se debe hacer un ajuste porque la nave está centrada en el túnel para t = 0 #

# --------------------- Gráfica --------------------- #

plt.figure(figsize=(20,8)) # Tamaño de la figura #
plt.title("Gráfico de Minkowski para un nave que pasa por un túnel, en el sistema túnel (O)") # Título del gráfico #

plt.axhline(0, color='black', linewidth=1.2) # Línea de referencia ct = 0 # 
plt.axvline(0, color='black', linewidth=1.2) # Línea de referencia x = 0 # 
plt.plot(x_entrada_O, ct_O, label='Línea de Entrada Túnel', color='red', linestyle='--' ) # Curva con la línea de entrada #
plt.plot(x_salida_O, ct_O, label='Línea de Salida Túnel', color='blue', linestyle='--' ) # Curva con la línea de salida #
plt.plot(x_O, ct_punta_O, label='Línea de Punta Nave', color='orange' ) # Curva con la línea de punta #
plt.plot(x_O, ct_cola_O, label='Línea de Cola Nave', color='green' ) # Curva con la línea de cola #
plt.plot(x_O, ct_luz_positiva_O, label='Rectas de Luz ct = ±x', color='black',linestyle=':' ) # Curva de luz positiva #
plt.plot(x_O, ct_luz_negativa_O,color='black',linestyle=':'  ) # Curva de luz negativa #

plt.scatter(A_x_O, A_ct_O , color='black', s=20) # Punto A # 
plt.text(A_x_O - 75, A_ct_O + 8, f'A ({A_x_O},{A_ct_O})', fontsize=9, color='black')
plt.scatter(B_x_O, B_ct_O , color='black', s=20) # Punto B #
plt.text(B_x_O + 8, B_ct_O - 20, f'B ({B_x_O},{B_ct_O})', fontsize=9, color='black')
plt.scatter(P_x_O, P_ct_O , color='black', s=20) # Punto P # 
plt.text(P_x_O + 10, P_ct_O - 6, f'P ({P_x_O},{P_ct_O})', fontsize=9, color='black')
plt.scatter(Q_x_O, Q_ct_O , color='black', s=20) # Punto Q #
plt.text(Q_x_O - 75, Q_ct_O + 4, f'Q ({Q_x_O},{Q_ct_O})', fontsize=9, color='black')

plt.xlabel(" x (m)") # Título eje x #
plt.ylabel(" ct (m)") # Título eje y #
plt.xlim(-200, 350) # Límites del eje x #
plt.ylim(-250, 250) # Límites del eje y #
plt.gca().set_aspect('equal', adjustable='box') # Ejes con la escala igual #
plt.grid(True, linestyle='--', alpha=0.6) # Muestra la grilla #
plt.legend(loc='upper left') # Muestra las leyendas #
plt.show() # Muestra el gráfico #
```

## b) Trasformación y verificación

```python
# --------------------- Matriz Lamda --------------------- #

Λ = np.array([[γ, -γ*β], [-γ*β,γ]]) # Matriz con la transformación de Lorentz

# --------------------- Curvas Tranformadas --------------------- #

matriz_entrada_O = np.vstack([ct_O,x_entrada_O]) # Une las dos listas de la referencia del túnel, en una matriz de 2 filas x N columnas #
matriz_salida_O = np.vstack([ct_O,x_salida_O]) # Une las dos listas de la referencia del túnel, en una matriz de 2 filas x N columnas #
matriz_entrada_prima = Λ @ matriz_entrada_O  # Multiplicación para generar la matriz con los valores primados # 
matriz_salida_prima = Λ @ matriz_salida_O  # Multiplicación para generar la matriz con los valores primados # 
x_entrada_prima = matriz_entrada_prima[1, :].tolist() # Extrae el arreglo respectivo con los valores primados # 
ct_entrada_prima = matriz_entrada_prima[0, :].tolist() # Extrae el arreglo respectivo con los valores primados # 
x_salida_prima = matriz_salida_prima[1, :].tolist() # Extrae el arreglo respectivo con los valores primados # 
ct_salida_prima = matriz_salida_prima[0, :].tolist() # Extrae el arreglo respectivo con los valores primados # 

matriz_punta_O = np.vstack([ct_punta_O,x_O]) # Une las dos listas de la referencia del túnel, en una matriz de 2 filas x N columnas #
matriz_cola_O = np.vstack([ct_cola_O,x_O]) # Une las dos listas de la referencia del túnel, en una matriz de 2 filas x N columnas #
matriz_punta_prima = Λ @ matriz_punta_O  # Multiplicación para generar la matriz con los valores primados # 
matriz_cola_prima = Λ @ matriz_cola_O  # Multiplicación para generar la matriz con los valores primados # 
x_punta_prima = matriz_punta_prima[1, :].tolist() # Extrae el arreglo respectivo con los valores primados # 
ct_punta_prima = matriz_punta_prima[0, :].tolist() # Extrae el arreglo respectivo con los valores primados # 
x_cola_prima = matriz_cola_prima[1, :].tolist() # Extrae el arreglo respectivo con los valores primados # 
ct_cola_prima = matriz_cola_prima[0, :].tolist() # Extrae el arreglo respectivo con los valores primados # 

matriz_luz_positiva_O = np.vstack([ct_luz_positiva_O,x_O]) # Une las dos listas de la referencia del túnel, en una matriz de 2 filas x N columnas #
matriz_luz_negativa_O = np.vstack([ct_luz_negativa_O,x_O]) # Une las dos listas de la referencia del túnel, en una matriz de 2 filas x N columnas #
matriz_luz_positiva_prima = Λ @ matriz_luz_positiva_O  # Multiplicación para generar la matriz con los valores primados # 
matriz_luz_negativa_prima = Λ @ matriz_luz_negativa_O  # Multiplicación para generar la matriz con los valores primados # 
x_luz_positiva_prima = matriz_luz_positiva_prima[1, :].tolist() # Extrae el arreglo respectivo con los valores primados # 
ct_luz_positiva_prima = matriz_luz_positiva_prima[0, :].tolist() # Extrae el arreglo respectivo con los valores primados # 
x_luz_negativa_prima = matriz_luz_negativa_prima[1, :].tolist() # Extrae el arreglo respectivo con los valores primados # 
ct_luz_negativa_prima = matriz_luz_negativa_prima[0, :].tolist() # Extrae el arreglo respectivo con los valores primados # 

# --------------------- Puntos Tranformados --------------------- #

A_prima = Λ @ np.array([[A_ct_O],[A_x_O]]) # Matriz con el par ordenado de A en el sistema O´ #
A_x_prima = round(A_prima[1,0],1) # Valor x # 
A_ct_prima = round(A_prima[0,0],1)  # Valor ct #

B_prima = Λ @ np.array([[B_ct_O],[B_x_O]]) # Matriz con el par ordenado de B en el sistema O´ #
B_x_prima = round(B_prima[1,0],1) # Valor x # 
B_ct_prima = round(B_prima[0,0],1)  # Valor ct #

P_prima = Λ @ np.array([[P_ct_O],[P_x_O]]) # Matriz con el par ordenado de P en el sistema O´ #
P_x_prima = round(P_prima[1,0],1) # Valor x # 
P_ct_prima = round(P_prima[0,0],1)  # Valor ct #

Q_prima = Λ @ np.array([[Q_ct_O],[Q_x_O]]) # Matriz con el par ordenado de Q en el sistema O´ #
Q_x_prima = round(Q_prima[1,0],1) # Valor x # 
Q_ct_prima = round(Q_prima[0,0],1)  # Valor ct #


# --------------------- Eventos Originales --------------------- #

print("=== Eventos originales en el sistema O (x, ct) ===")
print(f"A : ({A_x_O}, {A_ct_O}) m")
print(f"B : ({B_x_O}, {B_ct_O}) m")
print(f"P : ({P_x_O}, {P_ct_O}) m")
print(f"Q : ({Q_x_O}, {Q_ct_O}) m")

# --------------------- Eventos Transformados --------------------- #

print("=== Eventos transformados al sistema O' (x', ct') ===")
print(f"A' : ({A_x_prima}, {A_ct_prima}) m")
print(f"B' : ({B_x_prima}, {B_ct_prima}) m")
print(f"P' : ({P_x_prima}, {P_ct_prima}) m")
print(f"Q' : ({Q_x_prima}, {Q_ct_prima}) m")

L_0_medida = round(x_punta_prima[0] - x_cola_prima[0], 1) # Aquí se puede usar cualquier par # 
print(f"\nVerificación longitud propia de la nave en O': {L_0_medida} m (Esperado: {L_0} m del enunciado, L_0)")
```

## Interpretación física 

### ¿Qué lámpara emite primero en O′? ¿Por qué puede cambiar el orden sin violar la causalidad? 

Para el observador en la nave, la lámpara a la salida del túnel se enciende primero. Como se puede apreciar de la Gráfica B, el evento B´ (200.0,-160.0) sucede antes (tiempo menor, negativo) que el evento A´(0.0,0.0). 

El orden se puede cambiar porque los eventos A´ y B´ son independientes. De forma visual (Gráfica B), el evento B´ queda fuera del cono de luz del evento A´ y viceversa. Una situación similar sucede en el sistema O (Gráfica A). 


### ¿Cómo resulve el análsis de estas gráficas la aparente contradicción entre el piloto y los observadores del túnel?

Lo que parece una contracción se puede explicar por la relatividad de la simultaneidad. Aunque los eventos A y B sean simultáneos en el sistema O, no necesariamente lo son en el sistema O´ que se mueve a una rápidez constante $v = 0.800c $ respecto al primero. 

Para el sistema O, la distancia del túnel es $ D_0 = 120\;m $ y la longitud de la nave es $L= L_0 / \gamma = 108\;m $. Con esto, la nave sí cabe dentro del túnel para un observador en O. En la Gráfica A, la distancia $L$ es el espacio entre las curvas verde (línea de cola) y naranja (línea de punta) para un valor específico $ct$. En el instante t=0 del sistema O, la nave está centrada en el túnel y la distancia $L$ es menor que la distancia ($D_O$) entre los puntos A y B. Por lo tanto, los observadores de O sí pueden ver a la nave completamente dentro del túnel en el momento t=0. 

En el sistema O´, la longitud de la nave es $L_0 = 180\;m $ y la distancia del túnel es $ D = D_0 / \gamma = 72\;m $. Así, el túnel es más corto que la nave. Para la segunda gráfica (B), $D$ es la distancia entre las curvas roja (línea de entrada) y azul (línea de salida) para un valor específico $ct´$. En ct´=0, la distancia $L_0$ (espacio entre curvas verde y naranja) es mayor que la distancia $D_O$. Con esto, los observadores de O´ perciben un túnel más corto que la propia nave. Para el piloto, ambas lámparas sí encienden cuando el extremo correspondiente de la nave está dentro del túnel, pero no al mismo tiempo (puntos A´ y B´).

Como resultado, en x=x´=t=t´=0, ambas afirmaciones son correctas para los observadores en sus respectivos sistemas de referencia. La relatividad de la simultaneidad permite que la nave se encuentre completamente dentro del túnel para el sistema O y que el túnel tenga una distancia menor que la misma nave para el sistema O´.

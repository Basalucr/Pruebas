## Type system (sistema de tipado)

- <span class="fragment"> En un lenguaje de programación un sistema de tipado es un conjunto de reglas que clasifica *valores*, *variable* y *expresiones* en tipos, </span>
- <span class="fragment"> además define cómo estos tipos pueden interactuar y cómo deben ser utilizados correctamente.    </span>
- <span class="fragment"> Python utiliza un sistema de **tipado fuerte, dinámico y admite tipado de pato**.</span>
- <span class="fragment"> **Tipado fuerte:** el intérprete no permite operaciones entre tipos incompatibles, a menos que se lleve a cabo una conversión explícita.</span>

```python
resultado = 'foo' + 5 # Esto genera error 
```

```python
resultado = 'foo' + str(5) # Esto en cambio si funciona 
```

- **Tipado dinámico**: El tipo de una variable se determina en tiempo de ejecución al asignarle un valor, permitiendo reasignar una variable a un tipo diferente

````python
var1 = 5
var2 = 2 * var1

print(var2, type(var2)) 

var1 = "Hola "
var2 = 2 * var1

print(var2, type(var2)) 
````

- **Tipado Pato (Duck Typing)**: Se basa en la filosofía: "Si camina como pato y grazna como pato, entonces es un pato".


- <span class="fragment"> La compatibilidad se determina por los métodos que posee el objeto al momento de usarse. </span> 

````python
# Duck Typing (tipado de pato)
def sumar_patos(pato1, pato2):
    """Acepta cualquier objeto que defina el operador +."""
    return pato1 + pato2
````

````python
print(sumar_patos(2, 3))  # Los ints implementan el método suma
````

````python
print(sumar_patos('quack ', 'quack'))  # Los strings también implementan el método suma
````

````python
print(sumar_patos([1, 2, 3], [4, 5, 6]))  # Las listas también
````

- Podemos crear nuestro propio tipo "Pato" a través de una clase.
- <span class="fragment"> Si agregamos el método `__add__`, este tipo podrá sumarse. </span>

````python
class Pato:
    def __init__(self, valor):
        self.valor = valor
    def __str__(self): # Esto método nos permite "imprimir" el objeto
        return "Pato(" + str(self.valor) + ")"
    def __add__(self, other): # Este método nos permite "sumar" el objeto
        return Pato(self.valor + other.valor)


print(sumar_patos(Pato(5), Pato(7)))
# print( Pato(5) - Pato(7) ) # No Funciona porque no definimos el método __sub__ para la clase Pato
````

## Identación

- <span class="fragment"> La identación es es el espacio en blanco (generalmente 4 espacios o un `tab`) que se coloca antes de una instrucción para indicarle al intérprete que esa línea pertenece a un bloque específico (como una función, un bucle o un condicional). </span>

````python
var1 = 3

# Bloque: declaración 'if'
if var1 > 5:
    print("Esto se imprime si la condición es cierta")  # Identado: dentro de la declaración 'if'

print("Esto siempre se imprime") # Sin identar: fuera del 'if'
````

## Flujos de control
 
- <span class="fragment"> Los flujos de control son estructuras lógicas que modifican el orden en que se ejecutan las instrucciones. </span>
- <span class="fragment"> permiten tomar decisiones, repetir tareas o abordar un procesos de foma distinta cuando se cumple cierto criterio. </span>
- <span class="fragment"> Los flujos de control combinan expresiones booleanas con bloques de código identados. </span>
- <span class="fragment"> Python ofrece tres familias principales: condicionales, bucles determinados (`for`) y bucles indeterminados (`while`). </span>
- <span class="fragment"> Estas estructuras se complementan con sentencias como `pass`, `break` y `continue` que afinan el recorrido dentro de cada bucle.</span>

### Bucle `for`

  
- <span class="fragment"> `for` itera sobre cualquier objeto iterable (listas, strings, diccionarios, rangos, etc). </span>
- <span class="fragment"> Cada iteración obtiene un elemento y ejecuta el bloque asociado. </span>
- <span class="fragment"> Se puede combinar con funciones integradas como `range` para crear secuencias numéricas o con `enumerate` para obtener índices .</span>

````python
# Suponer queremos construir el siguiente string 
# "suma cuadrados = 0^2 + 1^2 + 2^2 + 3^2 + 4^2 = 30"
````

````python
mensaje = "suma cuadrados = "
suma_cuadrados = 0
N = 5
````

````python
for n in range(N): # 0, 1, ..., N-1
    
    suma_cuadrados += n**2
    mensaje += f"{n}^2"
    
    if n < N - 1:
        mensaje += " + "
    else:
        mensaje += " = "

print(mensaje + str(suma_cuadrados))
````

````python
# Iteración sobre lista
# -> Queremos enumerar estudiantes del aula
aula = ['Ana', 'Luis', 'Mariana', 'Juan']
id_est = 0
````

````python
for estudiante in aula:
    print(f"id: {id_est} estudiante: {estudiante}")
    id_est += 1
````

````python
for id_est, estudiante in enumerate(aula):
    print(f"id: {id_est} estudiante: {estudiante}")
````

### Bucle `while`


- <span class="fragment"> El bucle `while` evalua una condicón al inicio. </span>

- <span class="fragment"> Si la condición es `True`, el bloque se repite. </span>
  
- <span class="fragment"> Si la condición es `False`, salimos del bucle. </span>

- <span class="fragment"> Resulta ideal cuando no conocemos de antemano cuántas iteraciones serán necesarias (por ejemplo, esperar hasta que converja un método numérico).  </span>
 
- <span class="fragment"> Cuidado: Se recomienda agregar una condición de corte para evitar ciclos infinitos. </span>


````python
energia = 10 # Energía disponible
pasos = 0 # Pasos dados
````

````python
# A como damos pasos se consume la energía
# Seguimos iterando hasta que se agote la energía
while energia > 0:
    consumo = 3 if pasos % 2 == 0 else 1
    energia -= consumo
    pasos += 1
    print(f"Paso {pasos}: queda {max(energia, 0)} unidades")

print('Fin: sin energía disponible')
````

### `continue`

- <span class="fragment"> `continue` detiene la iteración en curso, </span>
- <span class="fragment"> y lleva al intérprete de regreso al inicio del bucle para evaluar la condición otra vez. </span>

````python
# Datos de un instrumento: Los valores `None` son mediciones fallidas
serie_datos_crudos = [
    1.45e2, None, None, 3.57e2, -4.90e2, None, 9.71e1, 8.66e1, 2.53e2, None
]
````

````python
serie_datos_limpios = []
for dato in serie_datos_crudos:
    if dato is None:
        continue   
    serie_datos_limpios.append(dato)
    print(f"Dato limpio: {dato}")

print(f"\nDatos limpios: {serie_datos_limpios}")
````

### `break`

- <span class="fragment"> `break` termina el bucle actual en el acto. </span>

````python
serie_datos_limpios = [145.0, 357.0, -490.0, 97.1, 86.6, 253.0]
primer_valor_negativo = None
````

````python
# Si encontramos un dato negativo, ya podemos dejar de iterar
for i, dato in enumerate(serie_datos_limpios):
    print(f"dato {i}: {dato}")
    if dato >= 0:
        continue
    primer_valor_negativo = dato
    break

print(f"\nPrimer valor negativo: {primer_valor_negativo}")
````

### Mutabilidad e inmutabilidad

<span class="fragment"> En Python los objetos pertenecen a una de dos categorías dependiendo de como reaccionan a modificaciones:  </span>

- <span class="fragment"> **Objetos inmutables**:   </span>
   - <span class="fragment"> no pueden cambiar su contenido una vez creados,  </span>
   - <span class="fragment"> una "modificación" realmente crea un objeto nuevo.  </span>
   - <span class="fragment"> Ejemplos: `int`, `float`, `bool`, `str`, `tuple`.  </span>

- <span class="fragment"> **Objetos mutables**:   </span>
   - <span class="fragment"> su contenido puede cambiar después de crearse,   </span>
   - <span class="fragment"> este cambio ocurre en el mismo lugar (**in-place**).  </span>
   - <span class="fragment"> Ejemplos: `list`, `dict`, `set`.  </span>

<span class="fragment"> **Notas**: </span>
- <span class="fragment"> Los objetos que pueden usarse para almacenar muchos datos suelen ser mutables porque copiar estructuras grandes es muy costoso en términos de memoria y computo.  </span>
- <span class="fragment"> Una excepción son las tuplas, ya que se usan para representar colecciones que no cambian: por ejemplo la colección de provincias de CR.  </span>

#### Parte 3: Copia, shallow copy y deep copy

- <span class="fragment"> `b = a` no copia: crea un alias (misma lista en memoria).  </span>
- <span class="fragment"> Shallow copy: nueva lista, pero comparte elementos internos si son mutables (`a.copy()`).  </span>
- <span class="fragment"> Deep copy: copia tambien las estructuras anidadas (`copy.deepcopy`).  </span>


````python
import copy

# 1) Aliasing
a = [1, 2, 3]

b = a

b.append(4)

print('a        :', a)
print('b (alias):', b)
````

````python
# 2) Copia superficial (shallow copy)
a = [1, 2, 3]

b = a.copy()

b.append(4)

print('a       :', a)
print('b (copy):', b)
````

````python
# 2) Copia superficial (shallow copy)
a = [[1, 2], [3, 4]] # Lista tiene elementos mutables

b = a.copy()

b.append([5,6])

print('a       :', a)
print('b (copy):', b)
````

````python
# 2) Shallow copy -> estructura nueva pero elementos internos mutables son compartidos
a[0].append(99)
a[1].append(-1)

print('a       :', a)
print('b (copy):', b)
````

````python
# 3) Deep copy (elementos internos son copiados recursivamente)
a = [[1, 2], [3, 4]]

b = copy.deepcopy(a)

b.append([5,6])
b[0].append(99)
b[1].append(-1)

print('a            :', a)
print('b (deep copy):', b)
````

#### Parte 4: Comprensión de listas (List comprehensions)

- <span class="fragment"> Forma general: `[expresion for x in iterable if condicion]`. </span>
- <span class="fragment"> Son útiles para transformar y filtar datos de forma rápida.  </span>
- <span class="fragment"> Son una forma elegante de construir series según un patrón. </span>

````python
mediciones_m = [0.12, None, 0.15, 0.11, None, 0.10]
````

````python
limpias_m = [m for m in mediciones_m if m is not None] # Filtrado
en_cm = [m * 100 for m in limpias_m] # Transformación

print('mediciones_m:', mediciones_m)
print('limpias_m:', limpias_m)
print('en_cm:', en_cm)
````

````python
N = 10
cuadrados_pares = [n**2 for n in range(N) if n % 2 == 0] # Patrón: cuadrados de pares
print('cuadrados_pares:', cuadrados_pares)
````

#### Parte 5:  Comprensión de listas anidadas y condicionales

## Funciones

- <span class="fragment"> Una función es una estructura que agrupa instrucciones utilizadas con frecuencia, bajo un nombre reutilizable.   </span>
- <span class="fragment"> Recibe argumentos y devuelve resultados mediante `return`.   </span>
- <span class="fragment"> Por qué usar funciones? reducen duplicidad, mejoran la legibilidad y permite pruebas unitarias aisladas.  </span>

### Argumentos Posicionales
 
- <span class="fragment"> Los argumentos se pasan en el orden exacto en que se definieron. </span>
- <span class="fragment"> Si se olvida uno, Python lanzará un error. </span>

````python
def sumar_patos_v1(pato1, pato2):
    return pato1 + pato2

print(sumar_patos_v1(5, 3))
# print(sumar_patos_v1(5)) # Error! 
````

### Argumentos por Defecto

- <span class="fragment"> Si el usuario no proporciona un valor, la función ya sabe qué usar.  </span>    

````python
def sumar_patos_v2(pato1, pato2=0):
    return pato1 + pato2

print(sumar_patos_v2(10))
print(sumar_patos_v2(10, 2))
````

### Argumentos Variables (*args)

- <span class="fragment"> Sirve para pasar un número indefinido de parámetros.  </span>
- <span class="fragment"> `*args` empaqueta todos los argumentos como una tupla.  </span>

````python
def sumar_patos_v3(*patos):
    suma = 0
    for pato in patos:
        suma += pato
    return suma

print(sumar_patos_v3(1, 2))
print(sumar_patos_v3(1, 2, 3, 4, 5))
````

### Argumentos por diccionario (**kwargs)

- <span class="fragment"> `**kwargs` (keyword arguments) permite pasar un número ilimitado de argumentos con nombre.  </span>   
- <span class="fragment"> Internamente la función recibe un diccionario.  </span>   
- <span class="fragment"> Es ideal para pasar configuraciones o metadatos. </span>  

````python
def sumar_patos_v4(pato1, pato2, **detalles_pato):
    print(detalles_pato, "\n")
    
    resultado = pato1 + pato2
    
    # Podemos acceder a argumentos extra por su nombre
    if detalles_pato.get("vuela"):
        print("¡Este resultado alcanzó el cielo!")
    return resultado

print(sumar_patos_v4(5, 5, vuela=True, color='amarillo', especie='pato doméstico'))
````

### Argmentos con Tipado

- <span class="fragment">Usamos typing para declarar qué esperamos recibir y qué vamos a devolver.  </span>
- <span class="fragment">No genera errores, pero ayuda a otros programadores a usar la función correctamente a no cometer errores.  </span>           

````python
def sumar_pato_v5(pato1: int, pato2: int) -> int:
    """Suma dos patos enteros y devuelve un pato entero."""
    return pato1 + pato2

print(sumar_pato_v5(10, 20))
print(sumar_pato_v5("quack ", "quack")) # Uso posible pero FUERA de la inteción original 
````

## Funciones que reciben funciones

- <span class="fragment"> Las funciones pueden recibir funciones como argumento.  </span>       

````python
def sumar_patos(pato1, pato2):
    return pato1 + pato2

def restar_patos(pato1, pato2):
    return pato1 - pato2

def aplicar_operacion(arg1, arg2, func):
    return func(arg1, arg2)

print(aplicar_operacion(4,6, sumar_patos))
print(aplicar_operacion(4,6, restar_patos))
````

## Funciones que devuelven funciones 

- <span class="fragment"> En Python las funciones pueden devolver funciones.  </span>      

````python
def generar_funcion(operacion="+"):
    if operacion == "+":
        def func(arg1, arg2):
            return arg1 + arg2
    elif operacion == "-":
        def func(arg1, arg2):
            return arg1 - arg2
    elif operacion == "*":
        def func(arg1, arg2):
            return arg1 * arg2
    elif operacion == "/":
        def func(arg1, arg2):
            return arg1 / arg2
    else:
        print("Operación desconocida!")
        def func(arg1, arg2):
            return None
        
    return func

multiplicar_patos = generar_funcion("*")
print(multiplicar_patos(5,7))
````

## Decoradores (Decorators)


- <span class="fragment"> Un decorador es una función que recibe otra función como argumento y extiende su comportamiento.     </span>
- <span class="fragment"> Se usa la notación @ para aplicar un decorador de forma limpia y legible.  </span>
- <span class="fragment"> Ejemplos de uso: medir tiempos de ejecución, validar permisos, cachear resultados, etc.  </span>
    

````python
def mi_decorador(func):
    def wrapper():
        print("--- Inicio de la ejecución ---")
        func()
        print("--- Fin de la ejecución ---")
    return wrapper
````

````python
@mi_decorador
def saludar():
    print("¡Hola, mundo!")

saludar()
````

## Uso útil de decoradores: profiling

````python
import time

def medir_tiempo(func):
    def wrapper(*args, **kwargs):

        inicio = time.perf_counter()
        resultado = func(*args, **kwargs)
        fin = time.perf_counter()
        
        tiempo_total = fin - inicio
        
        # Esto se puede enviar log file para análisis posterior
        print(f"Función '{func.__name__}' ejecutada en {tiempo_total:.4f} segundos") 
        
        return resultado
    return wrapper
````

````python
@medir_tiempo
def calcular_cuadrados(n):
    return [i**2 for i in range(n)]

# Probamos con cien mil elementos
mi_lista = calcular_cuadrados(100000)

# Probamos con un millón de elementos
mi_lista = calcular_cuadrados(1000000)
````

### Manejo de excepciones

- <span class="fragment"> Una **excepcion** interrumpe el flujo cuando ocurre un error (p. ej. `ZeroDivisionError`, `ValueError`, `FileNotFoundError`).  </span>
- <span class="fragment"> `try`: bloque donde *puede* ocurrir el error.  </span>
- <span class="fragment"> `except TipoError as e`: captura una excepcion especifica y permite reaccionar (log, mensaje, valor alternativo).  </span>
- <span class="fragment"> `else`: se ejecuta solo si **no** ocurrio ninguna excepcion en el `try`.  </span>
- <span class="fragment"> `finally`: se ejecuta **siempre** (ideal para limpieza/cierre de recursos).  </span>
- <span class="fragment"> `raise`: permite lanzar una excepcion (validacion) o relanzarla con mas contexto.  </span>
- <span class="fragment"> Recomendacion: captura excepciones **especificas** y evita `except:` salvo casos muy controlados.  </span>


#### Ejemplo 1: `try/except/else` (division segura)

- <span class="fragment"> Captura `ZeroDivisionError` y devuelve un valor alternativo.  </span>


````python
def dividir_seguro(numerador, denominador):
    try:
        resultado = numerador / denominador
    except ZeroDivisionError as e:
        print('No se puede dividir entre cero:', e)
        return None
    else:
        return resultado

print('10/2 =', dividir_seguro(10, 2))
print('10/0 =', dividir_seguro(10, 0))
````

#### Ejemplo 2: `try/except/finally` (recursos / archivos)

- <span class="fragment"> `finally` se ejecuta siempre, haya error o no.  </span>


````python
ruta = 'archivo_que_no_existe.txt'
try:
    with open(ruta, 'r', encoding='utf-8') as f:
        contenido = f.read()
        print('Contenido:', contenido[:50])
except FileNotFoundError as e:
    print('No encontre el archivo:', e)
finally:
    print('finally: aqui iria limpieza (cerrar conexiones, liberar recursos, etc.)')
````

#### Ejemplo 3: `raise` para validacion (y manejo aguas arriba)

- <span class="fragment"> Valida entradas y lanza `ValueError` con un mensaje claro.  </span>


````python
def energia_cinetica(masa, velocidad):
    if masa < 0:
        raise ValueError('La masa debe ser >= 0')
    return 0.5 * masa * velocidad**2

for m in [2.0, -1.0]:
    try:
        print('m=', m, 'Ec=', energia_cinetica(m, 3.0))
    except ValueError as e:
        print('Entrada invalida:', e)
````
